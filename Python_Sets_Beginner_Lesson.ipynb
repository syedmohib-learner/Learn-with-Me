## 1. What is a Set?

A **set** is a collection of items that stores **unique values**. This means duplicate values are removed automatically.

Important points:
- Sets do not keep duplicate values.
- Sets are unordered, so you should not rely on item positions or indexes.
- Sets can contain different data types, such as integers and strings.
- Set items must be immutable values, such as numbers, strings, or tuples containing immutable values.

```python
numbers = {1, 2, 3, 2, 1}
print(numbers)
```

**Explanation:** Although we wrote `1` and `2` more than once, each value appears only once in the set. The order shown when printed may vary.

## 2. Creating Sets

Use curly braces `{}` with values to create a set. To create an empty set, use `set()`.

**Remember:** `{}` creates an empty dictionary, not an empty set.

```python
fruits = {"apple", "banana", "mango"}
print(fruits)

empty_set = set()
print(empty_set)

not_a_set = {}
print(type(not_a_set))
```

## 3. Duplicate Values

A set automatically removes duplicates. This is useful when you want only distinct values.

```python
colors = {"red", "blue", "red", "green", "blue"}
print(colors)
print(len(colors))
```

**Explanation:** `len()` tells us how many unique items are in the set. The exact display order can differ.

## 4. Adding Items

Use `add()` to add **one** item. Use `update()` to add **multiple** items from an iterable, such as a list.

```python
animals = {"cat", "dog"}
animals.add("rabbit")
print(animals)

animals.update(["horse", "cow"])
print(animals)
```

## 5. Removing Items

- `remove(item)` removes an item, but causes an error if it is not present.
- `discard(item)` removes an item if present and does not cause an error if it is missing.
- `pop()` removes and returns an arbitrary item because sets are unordered.
- `clear()` removes all items.

```python
pets = {"cat", "dog", "fish"}
pets.remove("dog")
print(pets)

pets.discard("parrot")  # No error if "parrot" is absent
print(pets)

pets.clear()
print(pets)
```

## 6. Checking Whether an Item Exists

Use `in` to check if an item exists in a set. It gives `True` or `False`.

```python
subjects = {"English", "Python", "Math"}
print("Python" in subjects)
print("Biology" in subjects)
```

## 7. Set Operations

Suppose we have two sets:
- `A = {1, 2, 3}`
- `B = {3, 4, 5}`

**Union (`|`)**: all unique items from both sets.

**Intersection (`&`)**: items present in both sets.

**Difference (`-`)**: items in the first set but not in the second.

**Symmetric difference (`^`)**: items in either set, but not in both.

```python
A = {1, 2, 3}
B = {3, 4, 5}

print("Union:", A | B)
print("Intersection:", A & B)
print("Difference A - B:", A - B)
print("Symmetric difference:", A ^ B)
```

### Same operations using methods

You can also write these operations using methods:

```python
print(A.union(B))
print(A.intersection(B))
print(A.difference(B))
print(A.symmetric_difference(B))
```

## 8. Subsets and Supersets

`issubset()` checks whether every item in one set is also in another set.

`issuperset()` checks whether a set contains every item from another set.

```python
small = {1, 2}
large = {1, 2, 3, 4}

print(small.issubset(large))
print(large.issuperset(small))
```

## 9. Converting a List into a Set

Use `set()` to remove duplicates from a list. The order is not guaranteed.

```python
marks = [70, 80, 70, 90, 80, 100]
unique_marks = set(marks)
print(unique_marks)
```

## 10. Set Comprehension

A set comprehension is a short way to create a set using a loop-like expression.

This example creates the squares of numbers from 1 to 5. `range(1, 6)` generates 1, 2, 3, 4, 5.

```python
squares = {number ** 2 for number in range(1, 6)}
print(squares)
```

## 11. Set vs List vs Tuple

| Feature | Set | List | Tuple |
|---|---|---|---|
| Brackets | `{}` (with items) | `[]` | `()` |
| Duplicates | Not kept | Allowed | Allowed |
| Ordered | No guaranteed order | Yes | Yes |
| Indexing | No | Yes | Yes |
| Changeable | Yes | Yes | No |

Use a set when you need unique items or want to compare collections using set operations.

## 12. Common Mistakes

1. **Using `{}` for an empty set:** use `set()` instead.
2. **Trying to use an index:** `my_set[0]` does not work because sets do not support indexing.
3. **Expecting duplicates:** a set stores each unique value only once.
4. **Removing a missing item with `remove()`:** this raises an error; use `discard()` if the item may be absent.
5. **Expecting a fixed order:** sets are unordered.
