# Python Tuples

## 1. What is a Tuple?

A **tuple** is a collection of multiple values stored in a single variable.

### Simple Definition

> A tuple is an ordered collection of elements that cannot be changed after it is created.

### Example

````python
student = ("Ali", 20, "Computer Science")

print(student)
````

### Output

```text
('Ali', 20, 'Computer Science')
```

Here, the tuple contains three elements:

```text
"Ali"              → Name
20                 → Age
"Computer Science" → Department
```


# 2. Why Do We Use Tuples?

Tuples are useful when we have data that should **stay unchanged**.

For example:

````python
days = ("Monday", "Tuesday", "Wednesday", "Thursday", "Friday")
````

We normally don't want someone to accidentally change the days stored in this collection.

Other examples:

````python
coordinates = (10, 20)
rgb = (255, 128, 0)
student_record = ("Mohib", 21, "CS")
````

So a tuple can contain:

# 4. Tuple with One Element

This is a very important point.

If we write:

````python
x = (10)
````

Python considers this an integer:

````python
print(type(x))
````

Output:

```text
<class 'int'>
```

To create a tuple containing **one element**, we need a comma:

````python
x = (10,)
````

Now:

````python
print(type(x))
````

Output:

```text
<class 'tuple'>
```

### Remember

````python
(10)    # integer
(10,)   # tuple
````

The **comma** is what makes it a single-element tuple.

# 6. Accessing Tuple Elements

Tuples use **index numbers** to access their elements.

Indexing starts from `0`.

Example:

````python
fruits = ("Apple", "Banana", "Mango", "Orange")
````

Index positions:

```text
Apple   → 0
Banana  → 1
Mango   → 2
Orange  → 3
```

We can access an element using:

````python
print(fruits[0])
````

Output:

```text
Apple
```

Another example:

````python
print(fruits[2])
````

Output:

```text
Mango
```


# 7. Negative Indexing

Python also supports negative indexes.

The last element has index `-1`.

```text
Apple   → -4
Banana  → -3
Mango   → -2
Orange  → -1
```

Example:

````python
fruits = ("Apple", "Banana", "Mango", "Orange")

print(fruits[-1])
````

Output:

```text
Orange
```

And:

````python
print(fruits[-2])
````

Output:

```text
Mango
```

### Easy Rule

```text
Positive indexing → starts from 0
Negative indexing → starts from -1
```


# 8. Tuple Slicing

Just like strings and lists, tuples support slicing.

Syntax:

```python
tuple[start:end]
```

The `start` index is included, but the `end` index is not included.

Example:

````python
numbers = (10, 20, 30, 40, 50)

print(numbers[1:4])
````

Output:

```text
(20, 30, 40)
```

Why?

```text
Index:     0    1    2    3    4
Value:    10   20   30   40   50
                ↑         ↑
              start      end
```

Indexes `1`, `2`, and `3` are included.


# 9. Slicing from the Beginning

````python
numbers = (10, 20, 30, 40, 50)

print(numbers[:3])
````

Output:

```text
(10, 20, 30)
```

This means:

````python
numbers[0:3]
````

# 10. Slicing to the End

````python
numbers = (10, 20, 30, 40, 50)

print(numbers[2:])
````

Output:

```text
(30, 40, 50)
```


# 11. Slicing with a Step

We can also specify a step:

```python
tuple[start:end:step]
```

Example:

````python
numbers = (10, 20, 30, 40, 50, 60)

print(numbers[0:6:2])
````

Output:

```text
(10, 30, 50)
```

It selects every second element.


# 12. Tuples are Immutable

This is the **most important property of tuples**.

Immutable means:

> Once a tuple is created, its existing elements cannot be changed.

Example:

````python
try:
    numbers = (10, 20, 30)

    numbers[0] = 100
except Exception as e:
    print(type(e).__name__ + ":", e)
````

This produces an error:

```text
TypeError
```

Why?

Because tuples cannot be modified.

### Compare

List:

````python
numbers = [10, 20, 30]
numbers[0] = 100
````

This works.

Tuple:

````python
try:
    numbers = (10, 20, 30)
    numbers[0] = 100
except Exception as e:
    print(type(e).__name__ + ":", e)
````

This does not work.

# 18. Tuple Methods

Because tuples are immutable, they have fewer methods than lists.

The two important tuple methods are:

```text
count()

```


# 19. `count()` Method

`count()` tells us how many times a particular value appears.

Example:

````python
numbers = (10, 20, 10, 30, 10)

print(numbers.count(10))
````

Output:

```text
3
```

There are three `10`s.

Another example:

````python
fruits = ("Apple", "Banana", "Apple", "Mango")

print(fruits.count("Apple"))
````

Output:

```text
2
```


# 20. `index()` Method

`index()` tells us the index of the first occurrence of a value.

Example:

````python
fruits = ("Apple", "Banana", "Mango")

print(fruits.index("Banana"))
````

Output:

```text
1
```

Because:

```text
Apple  → 0
Banana → 1
Mango  → 2
```

If the value appears multiple times, `index()` returns the position of the **first occurrence**.

````python
numbers = (10, 20, 30, 20, 40)

print(numbers.index(20))
````

Output:

```text
1
```

# 24. Tuple Unpacking

Tuple unpacking is a very useful Python feature.

Suppose we have:

````python
student = ("Ali", 20, "CS")
````

We can assign its values to separate variables:

````python
name, age, department = student

print(name)
print(age)
print(department)
````

Output:

```text
Ali
20
CS
```

Python automatically matches:

```text
name       ← "Ali"
age        ← 20
department ← "CS"
```

### Important Rule

The number of variables should normally match the number of elements.

For example:

````python
try:
    student = ("Ali", 20, "CS")

    name, age = student
except Exception as e:
    print(type(e).__name__ + ":", e)
````

This causes an error because there are:

```text
3 values
2 variables
```


# 25. Swapping Variables Using Tuples

Tuple unpacking makes swapping values very easy.

Normally, beginners might use a temporary variable:

````python
a = 10
b = 20

temp = a
a = b
b = temp
````

Python allows:

````python
a = 10
b = 20

a, b = b, a

print(a)
print(b)
````

Output:

```text
20
10
```

Python internally handles this using tuple packing and unpacking.


# 26. Packing and Unpacking

### Packing

Putting multiple values into one tuple:

````python
student = "Ali", 20, "CS"
````

This is called **tuple packing**.

### Unpacking

Taking values from the tuple and assigning them to variables:

````python
name, age, department = student
````

This is called **tuple unpacking**.

### Easy way to remember

```text
Packing:
many values → one tuple

Unpacking:
one tuple → many variables
```

```text
Tuple → List → Modify → Tuple
```

# 34. Common Tuple Mistakes

## Mistake 1: Forgetting the comma

Wrong:

````python
x = (10)
````

This is an integer.

Correct:

````python
x = (10,)
````

## Mistake 2: Trying to change a tuple

````python
try:
    numbers = (10, 20, 30)

    numbers[0] = 100
except Exception as e:
    print(type(e).__name__ + ":", e)
````

This causes:

```text
TypeError
```

because tuples are immutable.


## Mistake 3: Wrong number of variables during unpacking

````python
try:
    student = ("Ali", 20, "CS")

    name, age = student
except Exception as e:
    print(type(e).__name__ + ":", e)
````

There are three values but only two variables.


## Mistake 4: Confusing `index()` with `count()`

````python
numbers = (10, 20, 10, 30)
````

`count()`:

````python
numbers.count(10)
````

means:

> How many times does 10 appear?

Answer:

```text
2
```

`index()`:

````python
numbers.index(10)
````

means:

> At what index does 10 first appear?

Answer:

```text
0
```

# 40. Final Summary

The easiest way to remember tuples is:

> **Tuple = ordered + immutable collection**
