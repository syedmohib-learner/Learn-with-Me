## 1. What Is a Dictionary?

**Definition:** A dictionary in Python is a collection of data stored in **key-value pairs**.

- **Key:** The name or identifier used to access data.
- **Value:** The data stored under that key.

Example:

| Key | Value |
|---|---|
| `name` | `Mohib` |
| `age` | `21` |
| `course` | `Python` |

Dictionaries use curly brackets `{}`. A colon `:` separates each key from its value, and commas separate pairs.

## 2. Creating a Dictionary

````python
student = {
    "name": "Mohib",
    "age": 21,
    "course": "Python"
}

print(student)
````

**Expected output**

```text
{'name': 'Mohib', 'age': 21, 'course': 'Python'}
```

## 3. Accessing Dictionary Values

Use a key inside square brackets `[]` to access its value.

````python
student = {
    "name": "Mohib",
    "age": 21,
    "course": "Python"
}

print(student["name"])
print(student["age"])
print(student["course"])
````

**Expected output**

```text
Mohib
21
Python
```

## 4. Adding New Items

````python
student = {
    "name": "Mohib",
    "age": 21
}

student["city"] = "Karachi"

print(student)
````

**Expected output**

```text
{'name': 'Mohib', 'age': 21, 'city': 'Karachi'}
```

Because `city` did not exist, Python adds a new key-value pair.

## 5. Updating Values

````python
student = {
    "name": "Mohib",
    "age": 21
}

student["age"] = 22

print(student)
````

**Expected output**

```text
{'name': 'Mohib', 'age': 22}
```

If a key already exists, assigning a new value updates it.

## 6. Removing Items

### Using `pop()`

````python
student = {
    "name": "Mohib",
    "age": 21,
    "city": "Karachi"
}

student.pop("age")

print(student)
````

**Expected output**

```text
{'name': 'Mohib', 'city': 'Karachi'}
```

### Using `del`

````python
student = {
    "name": "Mohib",
    "age": 21
}

del student["age"]

print(student)
````

**Expected output**

```text
{'name': 'Mohib'}
```

## 7. Important Dictionary Methods

| Method | Purpose |
|---|---|
| `keys()` | Returns all keys |
| `values()` | Returns all values |
| `items()` | Returns key-value pairs |
| `get()` | Gets a value using a key |
| `pop()` | Removes an item |
| `update()` | Adds or updates items |
| `clear()` | Removes all items |
| `copy()` | Creates a shallow copy |

### `keys()`

````python
student = {"name": "Mohib", "age": 21}
print(student.keys())
````

**Expected output:** `dict_keys(['name', 'age'])`

### `values()`

````python
student = {"name": "Mohib", "age": 21}
print(student.values())
````

**Expected output:** `dict_values(['Mohib', 21])`

### `items()`

````python
student = {"name": "Mohib", "age": 21}
print(student.items())
````

**Expected output:** `dict_items([('name', 'Mohib'), ('age', 21)])`

### `get()`

````python
student = {"name": "Mohib", "age": 21}

print(student.get("name"))
print(student.get("marks"))
print(student.get("marks", 0))
````

**Expected output**

```text
Mohib
None
0
```

If a key is missing, `get()` returns `None` by default. You can give it a default value, such as `0`.

### `update()`

````python
student = {"name": "Mohib", "age": 21}
student.update({"age": 22, "city": "Karachi"})
print(student)
````

**Expected output:** `{'name': 'Mohib', 'age': 22, 'city': 'Karachi'}`

## 8. Looping Through a Dictionary

### Loop through keys

````python
student = {
    "name": "Mohib",
    "age": 21,
    "course": "Python"
}

for key in student:
    print(key)
````

**Expected output**

```text
name
age
course
```

### Loop through values

````python
for value in student.values():
    print(value)
````

**Expected output**

```text
Mohib
21
Python
```

### Loop through keys and values

````python
for key, value in student.items():
    print(key, ":", value)
````

**Expected output**

```text
name : Mohib
age : 21
course : Python
```

`key` holds the current key, and `value` holds its corresponding value.

## 9. Checking Whether a Key Exists

````python
student = {
    "name": "Mohib",
    "age": 21
}

print("name" in student)
print("marks" in student)
````

**Expected output**

```text
True
False
```

The `in` operator checks keys, not values.

## 10. Dictionary Rules and Properties

- Dictionaries use curly brackets `{}`.
- Data is stored in key-value pairs.
- Keys must be unique.
- Values can be duplicated.
- Dictionaries are mutable: you can add, update, and remove items.
- Keys must be hashable. Common choices include strings, numbers, and suitable tuples.
- Values can have different data types, including lists and other dictionaries.
- Modern Python dictionaries preserve insertion order.

### Different Data Types

````python
data = {
    "name": "Mohib",
    "marks": 85,
    "passed": True,
    "subjects": ["Python", "AI", "Database"]
}

print(data)
````

## 11. Nested Dictionary

A nested dictionary is a dictionary inside another dictionary.

````python
students = {
    "student1": {
        "name": "Ali",
        "marks": 85
    },
    "student2": {
        "name": "Ahmed",
        "marks": 90
    }
}

print(students["student1"]["name"])
print(students["student2"]["marks"])
````

**Expected output**

```text
Ali
90
```

For example, `students["student1"]["name"]` first selects student1, then accesses that student's name.

## 12. Practical Example — Student Marks

````python
marks = {
    "Ali": 80,
    "Ahmed": 75,
    "Sara": 90
}

total = 0

for mark in marks.values():
    total = total + mark

print("Total marks:", total)
print("Number of students:", len(marks))
````

**Expected output**

```text
Total marks: 245
Number of students: 3
```

`len(marks)` returns the number of key-value pairs in the dictionary.
